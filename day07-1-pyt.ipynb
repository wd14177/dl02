```python
import time
import torch

def print_device_info():
    print("========== Device Info ==========")
    print(f"PyTorch version : {torch.__version__}")
    print(f"CUDA available  : {torch.cuda.is_available()}")

    if not torch.cuda.is_available():
        return

    device_idx = torch.cuda.current_device()
    props = torch.cuda.get_device_properties(device_idx)

    print(f"Using device    : cuda:{device_idx}")
    print(f"Device name     : {props.name}")
    print(f"Compute capability : {props.major}.{props.minor}")
    print(f"Total VRAM      : {props.total_memory / (1024**3):.2f} GB")
    print(f"CUDA version    : {torch.version.cuda}")
    print("=================================\n")


@torch.no_grad()
def gpu_burn_exact(seconds=10.0,
                   dtype=torch.float16,
                   size=8192,
                   use_relu=True):

    if not torch.cuda.is_available():
        raise RuntimeError("CUDA를 사용할 수 없습니다.")

    device = torch.device("cuda")

    torch.backends.cuda.matmul.allow_tf32 = True
    torch.set_float32_matmul_precision("high")

    print_device_info()
    print(f"Matrix size     : {size} x {size}")
    print(f"Dtype           : {dtype}")
    print(f"Burn time       : {seconds:.2f} seconds")
    print(f"Extra op        : {'relu' if use_relu else 'none'}\n")

    a = torch.randn((size, size), device=device, dtype=dtype)
    b = torch.randn((size, size), device=device, dtype=dtype)

    # 워밍업
    c = a @ b
    if use_relu:
        c.relu_()
    torch.cuda.synchronize()

    start = time.perf_counter()
    total_matmuls = 0

    while True:
        c = a @ b
        if use_relu:
            c.relu_()

        torch.cuda.synchronize()
        total_matmuls += 1

        if time.perf_counter() - start >= seconds:
            break

    total_time = time.perf_counter() - start

    # -------------------------
    # FLOPs 계산
    # -------------------------

    matmul_flops = 2 * (size ** 3)

    relu_flops = size ** 2 if use_relu else 0

    flops_per_iter = matmul_flops + relu_flops

    total_flops = flops_per_iter * total_matmuls

    print("\n========== Summary ==========")
    print(f"Total time          : {total_time:.3f} sec")
    print(f"Total matmuls       : {total_matmuls}")
    print(f"FLOPs per matmul    : {matmul_flops / 1e12:.3f} TFLOPs")
    if use_relu:
        print(f"ReLU FLOPs per iter : {relu_flops / 1e9:.3f} GFLOPs")

    print(f"--------------------------------")
    print(f"Total FLOPs         : {total_flops / 1e12:.3f} TFLOPs")
    print(f"Achieved TFLOPS     : {(total_flops / total_time) / 1e12:.3f}")
    print("=============================")


if __name__ == "__main__":
    gpu_burn_exact(
        seconds=10.0,
        dtype=torch.float16,
        size=8192,
        use_relu=True
    )
```

```python
import os
import time
import torch


def print_cpu_info():
    print("========== CPU Info ==========")
    print(f"PyTorch version : {torch.__version__}")
    print(f"Using device    : cpu")
    print(f"Intra-op threads: {torch.get_num_threads()}")
    print(f"Inter-op threads: {torch.get_num_interop_threads()}")

    for k in ["OMP_NUM_THREADS", "MKL_NUM_THREADS", "OPENBLAS_NUM_THREADS"]:
        if k in os.environ:
            print(f"{k} = {os.environ[k]}")
    print("================================\n")


@torch.no_grad()
def cpu_burn_exact(seconds=10.0,
                   dtype=torch.float32,
                   size=2048,
                   intraop_threads=None,
                   use_relu=True):
    """
    CPU에서 정확히 seconds 동안 matmul 반복 수행.
    종료 후 총 FLOPs 및 평균 GFLOPS 출력.
    """

    if intraop_threads is not None:
        torch.set_num_threads(int(intraop_threads))

    print_cpu_info()
    print(f"Matrix size     : {size} x {size}")
    print(f"Dtype           : {dtype}")
    print(f"Burn time       : {seconds:.2f} seconds")
    print(f"Extra op        : {'relu' if use_relu else 'none'}\n")

    device = torch.device("cpu")

    a = torch.randn((size, size), device=device, dtype=dtype)
    b = torch.randn((size, size), device=device, dtype=dtype)

    # 워밍업
    c = a @ b
    if use_relu:
        c.relu_()

    start = time.perf_counter()
    total_matmuls = 0

    while True:
        c = a @ b
        if use_relu:
            c.relu_()

        total_matmuls += 1

        if time.perf_counter() - start >= seconds:
            break

    total_time = time.perf_counter() - start

    # -------------------------
    # FLOPs 계산
    # -------------------------

    matmul_flops = 2 * (size ** 3)
    relu_flops = size ** 2 if use_relu else 0
    flops_per_iter = matmul_flops + relu_flops
    total_flops = flops_per_iter * total_matmuls

    print("\n========== Summary ==========")
    print(f"Total time          : {total_time:.3f} sec")
    print(f"Total matmuls       : {total_matmuls}")
    print(f"FLOPs per matmul    : {matmul_flops / 1e9:.3f} GFLOPs")
    if use_relu:
        print(f"ReLU FLOPs per iter : {relu_flops / 1e9:.3f} GFLOPs")

    print(f"--------------------------------")
    print(f"Total FLOPs         : {total_flops / 1e12:.3f} TFLOPs")
    print(f"Achieved GFLOPS     : {(total_flops / total_time) / 1e9:.3f}")
    print("=============================")


if __name__ == "__main__":
    cpu_burn_exact(
        seconds=10.0,
        dtype=torch.float32,
        size=2048,
        intraop_threads=None,  # 예: 8로 고정하고 싶으면 8
        use_relu=True
    )
    
```
